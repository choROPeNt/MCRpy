```python
# Create sample data
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.linspace(0, 1, 50)  # Values to be mapped to colors

# Create a scatter plot with the cividis colormap
scatter = plt.scatter(x, y, c=colors, cmap='cividis')

# Add a colorbar to the right of the plot
cbar = plt.colorbar(scatter)

# Set the colorbar to have 8 ticks
cbar.set_ticks(np.linspace(0, 1, 6))
cbar.set_ticklabels([f'{i:.3f}' for i in np.linspace(0, 1, 6)])

# Label the colorbar
cbar.set_label('Color Scale')

# Show the plot
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot with Cividis Colormap')
plt.show()
```

```python
from matplotlib.colors import ListedColormap, BoundaryNorm

# Create sample data
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)  # Values to be mapped to colors

# Define the discrete colormap
cmap = plt.get_cmap('cividis', 8)  # Discrete colormap with 8 colors

# Define the boundaries and normalize the colors
bounds = np.linspace(0, 1, 9)  # 8 intervals will have 9 boundaries
norm = BoundaryNorm(bounds, cmap.N)

# Create a scatter plot with the discrete colormap
scatter = plt.scatter(x, y, c=colors, cmap=cmap, norm=norm)

# Add a colorbar to the right of the plot
cbar = plt.colorbar(scatter, ticks=np.linspace(0, 1, 9))
cbar.set_ticklabels([f'{i:.3f}' for i in np.linspace(0, 1, 9)])

# Label the colorbar
cbar.set_label('Color Scale')

# Show the plot
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot with Discrete Cividis Colormap')
plt.savefig('scatter_plot.pdf', format='pdf',dpi=300)
plt.show()
```
